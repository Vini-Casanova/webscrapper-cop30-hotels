```python
import re

def parse_currency(val):
    if pd.isna(val):
        return None
    # Keep digits and decimal separator
    cleaned = re.sub(r"[^\d.,]", "", str(val))
    cleaned = cleaned.replace(",", ".")  # unify decimal format
    try:
        return float(cleaned)
    except ValueError:
        return None
```

```python
import json
import pandas as pd
import re

# Load JSON file
with open("batch1.json", "r", encoding="utf-8") as f:
    data = json.load(f)

# Ensure it's a list
if isinstance(data, dict):
    data = [data]

rows = []
for obj in data:
    bed_info = [item.get("body") for item in obj.get("structuredContent", {}).get("primaryLine", [])]
    image_urls = [img.get("url") for img in obj.get("images", [])]

    # Extract final price from break_down
    final_price = None
    for item in obj.get("price", {}).get("break_down", []):
        if isinstance(item.get("description"), str) and "noites x R$" in item["description"]:
            match = re.search(r"R\$\s?([\d.,]+)", item["description"])
            if match:
                # Convert to float
                final_price = float(match.group(1).replace(".", "").replace(",", "."))
                break

    rows.append({
        "room_id": obj.get("room_id"),
        "name": obj.get("name"),
        "title": obj.get("title"),
        "price": final_price,
        "rating_value": obj.get("rating", {}).get("value"),
        "rating_reviewCount": obj.get("rating", {}).get("reviewCount"),
        "bed_info": "; ".join(bed_info),
        "image_urls": "; ".join(image_urls),
        "latitude": obj.get("coordinates", {}).get("latitude"),
        "longitude": obj.get("coordinates", {}).get("longitud")
    })

# Convert to DataFrame
df = pd.DataFrame(rows)

print(df)
df.to_csv("output.csv", index=False)
```
