```python
import plotly.express as px
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1 = pd.read_csv('output.csv')
df2 = pd.read_csv('output2.csv')
df = pd.concat([df1,df2])
```

```python
df.head()
```

```python
df[['room_info', 'bed_info_detail']] = df['bed_info'].str.split(';', expand=True)

df['room_info'] = df['room_info'].str.strip()
df['bed_info_detail'] = df['bed_info_detail'].str.strip()

df['full_price'] = df['price']*11
```

```python
print(f'Before dropping duplicates: {len(df)}')
df = df.drop_duplicates()
df = df.loc[~df['room_id'].duplicated()]
print(f'Before after duplicates: {len(df)}')
```

```python
df = df[['name','title','price','full_price','rating_value','rating_reviewCount','latitude','longitude','room_info','bed_info_detail']]
```

```python
df = df.rename({
    'title':'category',
    'bed_info_detail':'bed_info'
},axis=1)
```

```python
import plotly.express as px

fig = px.scatter_map(
    df,
    lat="latitude",
    lon="longitude",
    hover_name="name",
    hover_data=["price", "bed_info"],
    color="price",
    zoom=10,
    height=2600,
    width=2000
)

# Use an open Mapbox style (no token needed)
fig.update_layout(mapbox_style="open-street-map")

# Automatically fit all points
fig.update_layout(
    mapbox=dict(
        bounds=dict(
            west=df['longitude'].min(),
            east=df['longitude'].max(),
            south=df['latitude'].min(),
            north=df['latitude'].max(),
        )
    )
)

# Optional: adjust margins for clean PNG export
fig.update_layout(margin={"r":0,"t":0,"l":0,"b":0})

# Export to PNG
fig.write_image("scatter_map.png", scale=10)
```

```python
fig = px.scatter_map(
        df,
        lat='latitude',
        lon='longitude',
        color='price',        # Color markers by city
        hover_name='name',   # Display city name on hover
        zoom=10,
        title='Distribuição apartamentos COP30', width=1980, height=1080
    )

fig.show()
```
